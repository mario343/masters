```python
def build_model(base_model, calibrated=False):
    pipe = Pipeline([
        # ("scaler", StandardScaler()),
        ("clf", base_model)
    ])
    if not calibrated:
        return pipe


def run_supervised_pipeline(model, X_train, y_train, X_full):
    model.fit(X_train, y_train)
    train_score = model.score(X_train, y_train)

    if hasattr(model, "predict_proba"):
        print("pipipupu")
        P_A = model.predict_proba(X_full)[:, 1]
    else:
        print("pupupipi")
        scores = model.decision_function(X_full)
        P_A = 1.0 / (1.0 + np.exp(-scores))

    return train_score, P_A
```

```python
k0_ref = 4.77
k4_ref = 1.178870

k477 = X_full_k477[::100]
n0 = k477[:, 0]
n4 = k477[:, 3]

med0 = np.median(n0)
med4 = np.median(n4)


reweightedP = []
reweightedVar = []

for k0, k4 in zip(k_dense, k4_dense):

    wght = np.exp(
        (k0 - k0_ref) * (n0 - med0)
        - (k4 - k4_ref) * (n4 - med4)
    )

    norm = np.sum(wght)

    mean_P = np.sum(P_A * wght) / norm
    mean_P2 = np.sum((P_A**2) * wght) / norm

    var_P = mean_P2 - mean_P**2

    reweightedP.append([k0, mean_P])
    reweightedVar.append([k0, var_P])

reweightedP = np.array(reweightedP)
reweightedVar = np.array(reweightedVar)
```

```python
from scipy.interpolate import interp1d

# create smooth inverse function: P → k0
f_inv = interp1d(reweightedP[:,1], reweightedP[:,0], kind='linear')

k0_half = f_inv(0.5)

print(k0_half)
```
