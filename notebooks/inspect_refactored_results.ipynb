# Inspect refactored results

This notebook only reads canonical CSV files and produces compact comparison tables.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / "Results").exists():
    ROOT = ROOT.parent
RESULTS_ROOT = ROOT / "Results" / "REFactored"
summary_files = sorted(RESULTS_ROOT.rglob("*_summary.csv"))
summary_files
```

```python
if summary_files:
    summary = pd.concat((pd.read_csv(path) for path in summary_files), ignore_index=True)
    display(summary)
else:
    print("No refactored results found yet.")
```
