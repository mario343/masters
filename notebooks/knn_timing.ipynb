# KNN timing benchmark

KNN is retained here only as an implementation-time benchmark. It is not a separate scientific phase-transition result.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / 'phase_transitions').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from phase_transitions.config import TRANSITIONS
from phase_transitions.knn_timing import benchmark_knn
```

```python
TRANSITION_NAMES = ['AB', 'BCB', 'CA']
TRAIN_STRIDES = {'AB': 40, 'BCB': 40, 'CA': 20}
FULL_STRIDES = {'AB': 100, 'BCB': 100, 'CA': 150}
FEATURE_SET_NAMES = ('30', '20', '12')
ALGORITHMS = ('auto', 'kd_tree', 'ball_tree', 'brute')
NEIGHBORS = tuple(list(range(1, 52, 2)) + [61, 81, 101, 151, 201])
BLOCK_COUNTS = range(2, 101)
OUTPUT_PATH = ROOT / 'Results' / 'SUPERVISED' / 'KNN_TIMING_summary.csv'
RUN = False  # Set True only after reviewing the benchmark configuration.
```

```python
if RUN:
    rows = []
    for transition_name in TRANSITION_NAMES:
        rows.extend(benchmark_knn(
            TRANSITIONS[transition_name],
            feature_set_names=FEATURE_SET_NAMES,
            algorithms=ALGORITHMS,
            neighbors=NEIGHBORS,
            train_stride=TRAIN_STRIDES[transition_name],
            full_stride=FULL_STRIDES[transition_name],
            block_counts=BLOCK_COUNTS,
        ))
    timing = pd.DataFrame(rows)
    display(timing.head())
    if OUTPUT_PATH.exists():
        raise FileExistsError(f'Refusing to overwrite {OUTPUT_PATH}')
    OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
    timing.to_csv(OUTPUT_PATH, index=False)
    print(OUTPUT_PATH)
else:
    print('RUN=False: no benchmark was executed.')
```
