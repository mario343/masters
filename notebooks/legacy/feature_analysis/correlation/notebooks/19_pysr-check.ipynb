```python
lhs = X_full[:, 5]

rhs = np.maximum.reduce([
    X_full[:, 11],
    X_full[:, 23],
    X_full[:, 17],
    X_full[:, 29],
])

print(np.max(np.abs(lhs - rhs)))

check2 = np.maximum(
    np.maximum(X_full[:, 11], X_full[:, 29]),
    np.maximum(X_full[:, 23],
    X_full[:, 17])
)
print(X_full[lhs - check2 != 0])
```

```python
import itertools
import numpy as np

lhs = X_full[:, 5]

features = [11, 23, 17, 29]

for r in range(1, len(features) + 1):
    print(f"\n=== {r} feature(s) ===")

    for combo in itertools.combinations(features, r):

        rhs = np.maximum.reduce([
            X_full[:, idx] for idx in combo
        ])

        err = np.max(np.abs(lhs - rhs))

        print(f"{combo} -> max error = {err}")
```
